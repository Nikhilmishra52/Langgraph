```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
from langchain_core.chat_history import BaseChatMessageHistory,InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
```

```python
store={}
```

```python
def get_session_id(session_id):
    if session_id not in store:
        store[session_id]=InMemoryChatMessageHistory()
    return store[session_id]
```

```python
from dotenv import load_dotenv
load_dotenv()
import os
llm = ChatGoogleGenerativeAI(model='gemini-2.5-flash',api_key=os.environ['GEMINI_API_KEY'])
```

```python
config = {"configurable":{"session_id":"first_chat"}}
```

```python
model_with_memory=RunnableWithMessageHistory(llm,get_session_id)
```

```python
model_with_memory.invoke(("hi i am nikhil"),config=config).content
```

```python
model_with_memory.invoke(("tell me what is my name"),config=config).content
```

```python
model_with_memory.invoke(("my age is 22 years"),config=config).content
```

```python
model_with_memory.invoke(("what is my name and age"),config=config).content
```
