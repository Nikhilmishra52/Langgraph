```python
from langchain.agents import create_agent
from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper
```

```python
wikiapi=WikipediaAPIWrapper()
```

```python
wiki_tool = WikipediaQueryRun(api_wrapper=wikiapi)
```

```python
wiki_tool.name
```

```python
wiki_tool.description
```

```python
tools=[wiki_tool]
```

```python
from langchain.agents import create_agent
```

```python
from dotenv import load_dotenv
load_dotenv()
from langchain_google_genai import ChatGoogleGenerativeAI
import os
llm = ChatGoogleGenerativeAI(model='gemini-2.5-flash',api_key=os.environ['GEMINI_API_KEY'])
```

```python
agent= create_agent(llm,tools=tools)
```

```python
from langchain_classic.schema import HumanMessage

agent.invoke({
    "messages": [
        HumanMessage(content="tell me about narendra modi")
    ]
})
```
